```python
my_set = set()

my_set.add(20)
my_set.add(30)
my_set.add(40)

my_set
```

```python
my_list = [10,20,20,30,40]

set_from_list = set(my_list)

set_from_list
```

```python
#union

a = set({10,20,25})

b = set({25,30,35})

new_set = a.union(b)

new_set
```

```python
#intersection


a = set({10,20,25})

b = set({25,30,35})

set_after_intrsection = a.intersection(b)

set_after_intrsection
```

```python
#subset


a = set({10,20,25})

b = set({5,10,25,20})

result = a.issubset(b)

result
```

```python
#comprehension

a = set({10,20,25})
my_set_squared = { i*i for i in a if i*i > 100}

my_set_squared
```
