```python
#accessing list elements


my_list = [10,20,30,40,50]

print("Element at 3rd position ===>",my_list[2])

print("Element at last index ===>",my_list[-1])

```

```python
#slicing list

#get elements from 0 to 2 postions

print("elements from 0 to 3 ===>",my_list[0:3])

print("all elements ===>",my_list[:])

print("from the last position to 3rd ===>", my_list[-1:-4:-1])
```

```python
# list methods
"""
3. List Methods

    These are the main built-in methods available for Python lists.

        Method	    Purpose
        append()	Add one item at the end
        extend()	Add multiple items
        insert()	Add item at a specific position
        remove()	Remove a specific value
        pop()	    Remove and return an item
        clear()	    Remove all items
        index()	    Find position of an item
        count()	    Count occurrences
        sort()	    Sort the list
        reverse()	Reverse the list
        copy()	    Create a copy

4. append()

    Adds one item to the end.

        numbers = [10, 20, 30]
        numbers.append(40)
        print(numbers)

    Output:
        [10, 20, 30, 40]

    Important
        numbers.append([50, 60])
        print(numbers)

    Output:

        [10, 20, 30, 40, [50, 60]]

    append() adds the entire object as one element.
"""

#append

my_list.append(60)
print(my_list)


```

```python
#extend
my_list.extend([70,80,90])
print(my_list)
```

```python
#index
my_list.insert(1,111)
print(my_list)
```

```python
#remove
my_list.remove(60)
print(my_list)
```

```python
#pop
result = my_list.pop()
print(result)
print(my_list)
```

```python
#clear
my_list.clear()
my_list
```

```python
#index
my_list = [10, 111, 50, 60, 70, 80, 90, 60, 70, 80, 90, 70, 80]

target_index = my_list.index(111)
my_list.pop(target_index)

print(my_list)
```

```python
#count
my_list.count(60)
```

```python
#reverse
reversed = my_list.reverse()
print("originsl list ==>", my_list)
print("returns ===>", reversed)
```

```python
#copy
deep_copied_list = my_list.copy()
deep_copied_list
```

```python
my_list1 = [50, 10, 60, 70, 80, 90]
my_list1.sort()
my_list1
```

```python
#tuples

my_tuple = (10,20,30,20,40,50,60,90,100)

count_20_in_my_tuple = my_tuple.count(20)
count_20_in_my_tuple
```

```python
numbers =  (1, 2, 3, 4)

new_tuple = [i*i for i in numbers if (i*i) %2 ==0]

new_tuple
```

```python
#all

all(i<=4 for i in numbers)
```

```python
#enumerate
names =  ("Amit", "Rahul", "Priya")


for index,name in enumerate(names, start=1):
    print(f"{name} - {index}")
```

```python
#zip

names = ("Amit", "Rahul", "Priya")
marks = (80, 90, 85)

for x,y in zip(names, marks):
    print(f"name={x} & marks={y}")
```
