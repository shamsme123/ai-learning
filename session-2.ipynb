```python
#fibonacci using while loop

n = int(input("Enter the Number"))
#1 1 2 3 5 8 13
a = 1
b = 1
print(a,end=" ")
print(b,end=" ")
c = 0
while c < n:
    c = a+b
    print(c,end=" ")
    a = b
    b = c
```

```python
#Number Guessing Game
secret_number = 7
counter = 1
while True:
    guessed_number = int(input("Enter the Number"))
    if guessed_number == secret_number:
        print(f"You have guessed the correct number {guessed_number}")
        break
    print("Try Agin!")
    print(f"Attempt count - {counter}")
    print("Number between 5 and 9")
    counter = counter + 1
    if counter > 3:
        print("You are allowed only 3 attempts!")
        break

    
```
